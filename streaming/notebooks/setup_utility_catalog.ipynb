```python
%sql
CREATE CATALOG airspace_pulse
MANAGED LOCATION 's3://airspace-lakehouse/airspace-pulse/';
```

```python
%sql
CREATE SCHEMA airspace_pulse.bronze;
CREATE SCHEMA airspace_pulse.silver;
CREATE SCHEMA airspace_pulse.gold;
```

```python
%sql
CREATE TABLE IF NOT EXISTS airspace_pulse.bronze.flight_states (
    time BIGINT,
    flight_vector ARRAY<STRING>,
    kafka_offset BIGINT,
    kafka_timestamp TIMESTAMP
)
USING DELTA
```

```python
# catalog: airspace_pulse
# ├── schema: bronze
# │   └── table: flight_states
# ├── schema: silver
# │   ├── table: flight_states_enriched
# │   └── table: flight_anomalies
# └── schema: gold
#     ├── table: congestion_by_sector
#     └── table: anomaly_events
```

```python
%sql
CREATE TABLE IF NOT EXISTS airspace_pulse.silver.flight_states_enriched (
    time            TIMESTAMP,
    icao24          STRING,
    callsign        STRING,
    origin_country  STRING,
    time_position   TIMESTAMP,
    last_contact    TIMESTAMP,
    longitude       FLOAT,
    latitude        FLOAT,
    baro_altitude   FLOAT,
    on_ground       BOOLEAN,
    velocity        FLOAT,
    true_track      FLOAT,
    vertical_rate   FLOAT,
    sensors         ARRAY<INT>,
    geo_altitude    FLOAT,
    squawk          STRING,
    spi             BOOLEAN,
    position_source INT,
    category        INT,
    kafka_offset BIGINT,
    kafka_timestamp TIMESTAMP
)
USING DELTA
```

```python
%sql
CREATE TABLE IF NOT EXISTS airspace_pulse.silver.flight_states_anomalies (
    icao24          STRING, 
    callsign        STRING,
    event_time      TIMESTAMP, -- time when the anomaly occurred
    anomaly_type    STRING,
    severity        STRING,
    detail          STRING,
    longitude       FLOAT,
    latitude        FLOAT,
    detected_at     TIMESTAMP -- time when the anomaly was detected
)
USING DELTA
```
